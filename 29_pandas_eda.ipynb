```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
# Read the data
current_dir = Path.cwd()
file_path = current_dir / "data" /"house-rent-data-1"/"House_Rent_Dataset.csv"
df = pd.read_csv(file_path)
df.head()
```

```python
df.describe()
```

```python
df.columns
```

```python
# Average Rent based on city
avg_by_city = df.groupby("City")["Rent"].mean().sort_values(ascending=False)
avg_by_city
```

```python
# Plot of average by city
avg_by_city.plot(kind="bar", figsize=(10,6))
plt.title("Average Rent by City")
plt.xlabel("City")
plt.ylabel("Average Rent")
plt.show()
```

```python
# value counts for City
df["City"].value_counts()
```

```python
# Plot mean line on bargraph for rent vs city
avg_by_city.plot(kind="bar", figsize=(10,6))
plt.axhline(y=avg_by_city.mean(), color='r', linestyle='--')
plt.title("Average Rent by City with Mean Line")
plt.xlabel("City")
plt.ylabel("Average Rent")
plt.show()
```

```python
# Average rent by size
avg_by_size = df.groupby("Size")["Rent"].mean().sort_values(ascending=False)
avg_by_size

# Plot of average by size
avg_by_size.plot(kind="bar", figsize=(10,6))
plt.title("Average Rent by Size")
plt.xlabel("Size")
plt.ylabel("Average Rent")
plt.show()

# Plot mean line on bargraph for rent vs size
avg_by_size.plot(kind="bar", figsize=(10,6))
plt.axhline(y=avg_by_size.mean(), color='r', linestyle='--')
plt.title("Average Rent by Size with Mean Line")
plt.xlabel("Size")
plt.ylabel("Average Rent")
plt.show()
```

```python
# What is the average rent by number of bedrooms
avg_by_bedrooms = df.groupby("BHK")["Rent"].mean().sort_values(ascending=False)
avg_by_bedrooms

# Plot of average by number of bedrooms
avg_by_bedrooms.plot(kind="bar", figsize=(10,6))
plt.title("Average Rent by Number of Bedrooms")
plt.xlabel("Number of Bedrooms")
plt.ylabel("Average Rent")
plt.show()

# Plot mean line on bargraph for rent vs number of bedrooms
avg_by_bedrooms.plot(kind="bar", figsize=(10,6))
plt.axhline(y=avg_by_bedrooms.mean(), color='r', linestyle='--')
plt.title("Average Rent by Number of Bedrooms with Mean Line")
plt.xlabel("Number of Bedrooms")
plt.ylabel("Average Rent")
plt.show()
```

# Summary Insights

```python
# The city with the highest average rent
highest_avg_rent_city = avg_by_city.idxmax()
print(f"The city with the highest average rent is {highest_avg_rent_city}")

# The size with the highest average rent
highest_avg_rent_size = avg_by_size.idxmax()
print(f"The size with the highest average rent is {highest_avg_rent_size}")

# The number of bedrooms with the highest average rent
highest_avg_rent_bedrooms = avg_by_bedrooms.idxmax()
print(f"The number of bedrooms with the highest average rent is {highest_avg_rent_bedrooms}")
```
